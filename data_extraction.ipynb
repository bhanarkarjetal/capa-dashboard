# CAPA Dashboard - Data Extraction

## Data Source

Download the raw JSON from the [openFDA Device Enforcement dataset](https://open.fda.gov/data/downloads/).
Save the file as `device-recall-0001-of-0001.json` inside the `data/` folder before running this notebook.

```python
import json
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path('data')
```

```python
with open(DATA_DIR / 'device-recall-0001-of-0001.json', 'r') as f:
    raw = json.load(f)

df_raw = pd.DataFrame(raw['results'])
```

```python
print(df_raw.shape)
df_raw.head()
```

```python
infusion_df = df_raw[df_raw['product_description'].str.contains('infusion pump', case=False, na=False)]
```

```python
infusion_df.shape
```

```python
infusion_df.to_csv(DATA_DIR / 'infusion_pump_recalls.csv', index=False)
print('Saved successfully')
```
