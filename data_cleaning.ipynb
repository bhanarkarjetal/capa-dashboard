# CAPA Dashboard - Data Cleaning

Cleans and reclassifies FDA infusion pump recall data to produce `recall_df_clean.csv`.

**Input:** `data/infusion_pump_recalls.csv`: 666 infusion pump recall records from openFDA

**Output:** `data/recall_df_clean.csv`: 529 records with corrected root cause classifications

## Data Source

Source: [openFDA Device Enforcement dataset](https://open.fda.gov/data/downloads/).

The raw dataset (1.4 GB JSON) was filtered for infusion pump records using `data_extraction.ipynb`.

Download `infusion_pump_recalls.csv` directly from the README if you want to skip that step.

```python
# Load Data
from pathlib import Path
import pandas as pd
```

```python
DATA_DIR = Path('data')
```

```python
recall_df = pd.read_csv(DATA_DIR / 'infusion_pump_recalls.csv')
```

```python
# Explore
recall_df.head()
```

```python
recall_df.columns
```

```python
# Filter & Select
recall_df['corrected_root_cause'] = recall_df['root_cause_description']
print(recall_df['corrected_root_cause'].value_counts())
```

```python
recall_df = recall_df[recall_df['root_cause_description'] != 'Other']
print(recall_df.shape)
```

```python
columns_to_keep = [
    'cfres_id',
    'product_res_number',
    'recalling_firm',
    'product_description',
    'reason_for_recall',
    'root_cause_description',
    'corrected_root_cause',
    'recall_status',
    'event_date_initiated',
    'event_date_terminated',
    'action'
]

recall_df = recall_df[columns_to_keep]
print(recall_df.shape)
```

```python
recall_df.head()
```

```python
recall_df.to_csv(DATA_DIR / 'infusion_pump_recalls_clean.csv', index=False)
print('Saved successfully')
```
